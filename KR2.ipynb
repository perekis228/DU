```python
from sympy import *

t = symbols('t', positive=True)
y = Function('y')
y0 = 10
L = 18
g = Rational(98, 10)

eq1 = Eq(L*y(t).diff(t, 2), (2*y(t)-L)*g)
ics = {y(0): y0, y(t).diff(t).subs(t, 0): 0}
y_t = dsolve(eq1, y(t), ics=ics)

eq2 = Eq(y_t.rhs, L)
t = [sol.evalf() for sol in solve(eq2) if sol.is_real][0]
t
```
