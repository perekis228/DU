```python
from sympy import *
```

```python
m = 14520
F_c = m / 200 #Сила сопротивления
S = 5
g = 9.8
k = 0.12*g
v = 15 #Скорость ветра
w_0 = 0

t, C1 = symbols('t C1')
w = Function('w') #Скорость вагона
u = Function('u') #Скорость скорость ветра относительно вагона
F_d = Function('F_d') #Сила давления
h = Function('h')

eq_u = Eq(u(t), 15 - w(t))

eq_u
```

```python
eq_F_d = Eq(F_d(t), k*S*eq_u.rhs**2)
eq_F_d
```

```python
eq = Eq(m*Derivative(w(t), t), eq_F_d.rhs - F_c)
w_t_c = dsolve(eq, w(t))
w_t_c
```

```python
C1_val = nsolve(Eq(0, w_t_c.rhs).subs(t, 0), -100)
w_t = Eq(w(t), w_t_c.rhs).subs(C1, C1_val)
w_t
```

```python
plot(w_t.rhs, (t, 0, 5000))
```

```python
w_max = w_t.rhs.subs(t, 5000)
w_max
```

```python
t_do_3 = nsolve(Eq(3, w_t.rhs), 0)
t_do_3
```

```python
h_t = dsolve(Eq(Derivative(h(t), t), w_t.rhs), h(t))
h_t
```
